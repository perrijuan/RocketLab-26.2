```python



path_movies_info = f"{landing_path}/movies_info_TMDB_IMDB.csv"
path_movies_fincancials = f"{landing_path}/movies_financials_IMDB_TMDB.csv"
path_metics = f"{landing_path}/movies_metrics_IMDB_TMDB.csv"
path_credi_swiss = f"{landing_path}/credits_and_tags_IMDB_TMDB.csv"
path_review = f"{landing_path}/movies_reviews.csv"
# usei o codigo do grupo ele altera  a ordem do ano mes e dia !!!! 
df_analytics_dollar_pix = ingerir_ptax(spark, date(2026, 9, 1), date.today())


df_movie_info = spark.read.csv(path_movies_info, header=True, inferSchema=True)
df_movies_financials = spark.read.csv(path_movies_fincancials, header=True, inferSchema=True)
df_metrics = spark.read.csv(path_metics, header=True, inferSchema=True)
df_credi_swiss = spark.read.csv(path_credi_swiss, header=True, inferSchema=True)
df_review = spark.read.csv(path_review, header=True, inferSchema=True)

```

```python
# 3. Gravação com adição do timestamp no momento da escrita
df_movie_info \
    .withColumn("ingestion_datetime", current_timestamp()) \
    .write.format("delta").mode("Append") \
    .saveAsTable(f"{bronze_schema}.tb_movies_info")

df_movies_financials \
    .withColumn("ingestion_datetime", current_timestamp()) \
    .write.format("delta").mode("Append") \
    .saveAsTable(f"{bronze_schema}.tb_movies_financials")

df_metrics \
    .withColumn("ingestion_datetime", current_timestamp()) \
    .write.format("delta").mode("Append") \
    .saveAsTable(f"{bronze_schema}.tb_movies_metrics")

df_credi_swiss \
    .withColumn("ingestion_datetime", current_timestamp()) \
    .write.format("delta").mode("Append") \
    .saveAsTable(f"{bronze_schema}.tb_credits_and_tags")

df_review \
    .withColumn("ingestion_datetime", current_timestamp()) \
    .write.format("delta").mode("Append") \
    .saveAsTable(f"{bronze_schema}.tb_movies_reviews")

df_analytics_dollar_pix \
    .withColumn("ingestion_datetime", current_timestamp()) \
    .write.format("delta").mode("Append") \
    .saveAsTable(f"{bronze_schema}.tb_cotacao_dolar")

display(spark.table(f"{bronze_schema}.tb_movies_info").limit(5))
```
